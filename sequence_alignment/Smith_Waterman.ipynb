```python
import numpy as np
```

```python
seq1="ATGCT"
seq2="AGCT"
```

```python
matrix= np.zeros((len(seq1)+1,len(seq2)+1))
```

```python
match_matrix=np.zeros((len(seq1),len(seq2)))
```

```python
match=1
mismatch=-1
gap=-2
```

```python
#Just for check
```

```python
for i in range(len(seq1)):
  for j in range(len(seq2)):
    if seq1[i]==seq2[j]:
      match_matrix[i][j]=match
    else:
      match_matrix[i][j]=mismatch
```

```python
print(match_matrix)
```

```python
#Waterman Code
```

```python
# the first row and coloumn
```

```python
for i in range(len(seq1)+1):
  matrix[i][0]=gap*i
for j in range(len(seq2)+1):
  matrix[0][j]=gap*j
```

```python
#matrix fill
```

```python
for i in range(len(seq1)):
  for j in range(len(seq2)):
    matrix[i+1][j+1]=max(matrix[i][j+1]+gap,matrix[i+1][j]+gap,matrix[i][j]+match_matrix[i][j])
    if matrix[i+1][j+1]<0:
      matrix[i+1][j+1]=0

print(matrix)
```

```python
#trace back
```

```python
aligned_1=""
aligned_2=""
ti=len(seq1)
tj=len(seq2)
```

```python
while(ti >0 and tj > 0):
 if (ti >0 and tj > 0 and matrix[ti][tj] == matrix[ti-1][tj-1]+ match_matrix[ti-1][tj-1]):

        aligned_1 = seq1[ti-1] + aligned_1
        aligned_2 = seq2[tj-1] + aligned_2

        ti = ti - 1
        tj = tj - 1
 elif(ti > 0 and matrix[ti][tj] == matrix[ti-1][tj] + gap):
        aligned_1 = seq1[ti-1] + aligned_1
        aligned_2 = "-" + aligned_2

        ti = ti -1
 else:
        aligned_1 = "-" + aligned_1
        aligned_2 = seq2[tj-1] + aligned_2

        tj = tj - 1
```
